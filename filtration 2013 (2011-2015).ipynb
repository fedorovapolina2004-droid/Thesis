```python
import pandas as pd

df = pd.read_excel("dataset_final_new_new.xlsx")
df.columns = df.columns.astype(str).str.strip().str.lower()


df["year"] = pd.to_numeric(df["year"], errors="coerce")
df["sum"]  = pd.to_numeric(df["sum"], errors="coerce")


df_p = df[df["year"].between(2011, 2015)].copy()


yearly = (
    df_p.groupby(["brand", "model", "year"], as_index=False)["sum"]
    .sum()
)


ok_models = (
    yearly[yearly["sum"] > 0]
    .groupby(["brand", "model"])["year"]
    .nunique()
    .reset_index(name="n_years")
)

ok_models = ok_models[ok_models["n_years"] == 5][["brand", "model"]]


final_df = df.merge(ok_models, on=["brand", "model"], how="inner")
final_df = final_df[final_df["year"].between(2011, 2015)]

print("Моделей прошло фильтр (2011–2015):", ok_models.shape[0])
final_df.head()
```

```python
cc_check = (
    final_df.groupby(["brand", "model"])["cc"]
    .nunique()
    .reset_index(name="n_cc")
)


cc_problem = cc_check[cc_check["n_cc"] > 1]

print("Моделей с несколькими cc:", cc_problem.shape[0])
cc_problem.head()
```

```python
df["cc"] = pd.to_numeric(df["cc"], errors="coerce")
df["model_cc"] = df["model"] + " " + df["cc"].astype(int).astype(str) + "cc"

df_p = df[df["year"].between(2011, 2015)].copy()

yearly = (
    df_p.groupby(["brand", "model_cc", "year"], as_index=False)["sum"]
    .sum()
)

ok_models = (
    yearly[yearly["sum"] > 0]
    .groupby(["brand", "model_cc"])["year"]
    .nunique()
    .reset_index(name="n_years")
)

ok_models = ok_models[ok_models["n_years"] == 5][["brand", "model_cc"]]

final_df = df_p.merge(ok_models, on=["brand", "model_cc"], how="inner")

print("Количество версий моделей:", ok_models.shape[0])
final_df.head()
```

```python
final_df.to_csv(
    "new_dataset_2011_2015_model_split_by_cc.csv",
    index=False,
    encoding="utf-8-sig"
)
```

```python

df["cc"] = pd.to_numeric(df["cc"], errors="coerce")


final_df = df.merge(ok_models, on=["brand", "model"], how="inner")

final_df = final_df[final_df["year"].between(2011, 2015)]


final_df = final_df[(final_df["cc"] <= 1200) & (final_df["cc"] > 0)]

print("Моделей после фильтра cc ≤ 1200:",
      final_df[["brand", "model"]].drop_duplicates().shape[0])


final_df = final_df.sort_values(["brand", "model", "year", "month"])


final_df.to_csv(
    "filtered_2011_2015_cc_le_1200.csv",
    index=False,
    encoding="utf-8-sig"
)

print("CSV файл сохранён как filtered_2011_2015_cc_le_1200.csv")
```
