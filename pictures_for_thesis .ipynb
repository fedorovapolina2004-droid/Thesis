```python
import matplotlib.pyplot as plt
import textwrap

# ---------- Данные ----------
title = "Figure 11 - OLS regression, The 2018 Policy"
source = "Source: Compiled by the Author"

note = (
    "Note: This table presents the results of estimating Equation 3 using vehicle sales as the outcome variable. "
    "The estimation is performed in a panel at the brand-model level. The reported coefficients correspond to "
    "the interaction terms between the treatment indicator and year dummies, with 2017 omitted as the baseline year. "
    "The coefficients therefore show the difference in sales between treated and control vehicles relative to the baseline year. "
    "The model includes brand fixed effects and year fixed effects. Standard errors are reported in parentheses. "
    "*** p<0.01, ** p<0.05, * p<0.1."
)

coef_rows = [
    ("Treated x 2016", "-0.1218",  "(0.0887)"),
    ("Treated x 2018", "-0.1982",  "(0.1331)"),
    ("Treated x 2019", "-0.1625",  "(0.0995)"),
    ("Treated x 2020", "-0.2221*", "(0.1158)"),
    ("Treated x 2021", "-0.0814",  "(0.1445)"),
    ("Treated x 2022", "-0.3089*", "(0.1723)"),
]

summary_rows = [
    ("Observations", "23100"),
    ("Brand FE", "Yes"),
    ("Year FE", "Yes"),
]

# ---------- Настройка фигуры ----------
fig, ax = plt.subplots(figsize=(8.6, 8.2))
ax.axis("off")

# Позиции
left = 0.10
right = 0.90
x_var = 0.18
x_val = 0.73

# Верхняя линия
ax.plot([left, right], [0.94, 0.94], color="black", linewidth=1.6)

# Заголовок таблицы
ax.text(x_var, 0.885, "Variable name", fontsize=13, ha="left", va="center", family="serif")
ax.text(x_val, 0.90, "sales", fontsize=13, ha="center", va="center", family="serif")
ax.plot([x_val - 0.10, x_val + 0.10], [0.87, 0.87], color="gray", linewidth=1.0)
ax.text(x_val, 0.845, "(1)", fontsize=13, ha="center", va="center", family="serif")

# Линия под заголовком
ax.plot([left, right], [0.81, 0.81], color="black", linewidth=1.1)

# Основные коэффициенты
y = 0.77
step = 0.085

for var, coef, se in coef_rows:
    ax.text(x_var, y, var, fontsize=13, ha="left", va="center", family="serif")
    ax.text(x_val, y + 0.01, coef, fontsize=13, ha="center", va="center", family="serif")
    ax.text(x_val, y - 0.025, se, fontsize=13, ha="center", va="center", family="serif")
    y -= step

# Линия перед summary
ax.plot([left, right], [y + 0.03, y + 0.03], color="black", linewidth=1.1)

# Summary block
y_summary = y - 0.01
for label, value in summary_rows:
    ax.text(x_var, y_summary, label, fontsize=13, ha="left", va="center", family="serif")
    ax.text(x_val, y_summary, value, fontsize=13, ha="center", va="center", family="serif")
    y_summary -= 0.045

# Нижняя линия
bottom_line_y = y_summary + 0.015
ax.plot([left, right], [bottom_line_y, bottom_line_y], color="black", linewidth=1.6)

# Significance levels
ax.text(
    x_var, bottom_line_y - 0.03,
    "Significance levels:  *** p<0.01,  ** p<0.05,  *p<0.1",
    fontsize=10.5,
    ha="left",
    va="center",
    family="serif",
    style="italic"
)

# Caption
ax.text(
    left, bottom_line_y - 0.09,
    title,
    fontsize=11.5,
    ha="left",
    va="center",
    family="serif",
    fontweight="bold"
)

ax.text(
    left, bottom_line_y - 0.13,
    source,
    fontsize=10.5,
    ha="left",
    va="center",
    family="serif"
)

# Note с переносом строк
wrapped_note = textwrap.fill(note, width=118)
ax.text(
    left, bottom_line_y - 0.17,
    wrapped_note,
    fontsize=10,
    ha="left",
    va="top",
    family="serif",
    linespacing=1.35
)

plt.tight_layout()

# ---------- Сохранение ----------
plt.savefig("figure_11_ols_2018.png", dpi=300, bbox_inches="tight")
plt.savefig("figure_11_ols_2018.pdf", bbox_inches="tight")

plt.show()
```
