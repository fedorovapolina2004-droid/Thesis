```python
import pandas as pd

df = pd.read_excel("dataset_final_new_new.xlsx")
df.columns = df.columns.astype(str).str.strip().str.lower()


df["year"] = pd.to_numeric(df["year"], errors="coerce")
df["sum"]  = pd.to_numeric(df["sum"], errors="coerce")


df_p = df[df["year"].between(2016, 2020)].copy()


yearly = (
    df_p.groupby(["brand", "model", "year"], as_index=False)["sum"]
    .sum()
)


ok_models = (
    yearly[yearly["sum"] > 0]
    .groupby(["brand", "model"])["year"]
    .nunique()
    .reset_index(name="n_years")
)

ok_models = ok_models[ok_models["n_years"] == 5][["brand", "model"]]


final_df = df.merge(ok_models, on=["brand", "model"], how="inner")
final_df = final_df[final_df["year"].between(2016, 2020)]

print("Моделей прошло фильтр (2016–2020):", ok_models.shape[0])
final_df.head()
```

```python

final_df = final_df.sort_values(["brand", "model", "year", "month"])


final_df.to_csv(
    "filtered_models_2016_2020.csv",
    index=False,
    encoding="utf-8-sig"
)

print("Файл сохранён как filtered_models_2016_2020.csv")
```
